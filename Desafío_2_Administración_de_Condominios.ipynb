```python
import pandas as pd
```

```python
condominio=pd.read_json('/content/Condominios.json')
condominio.head()
```

```python
condominio=pd.json_normalize(condominio['dados_locacao'])
condominio.head()
```

Eliminar datos en listas dentro del DataFrame;

Verificar tipos de datos;

Identificar columnas numéricas;

Transformar la columna numérica a tipo numérico.

```python
columnas=list[condominio.columns]
columnas
```

```python
condominio = condominio.explode(condominio.columns.tolist()[1:])
condominio.head()
```

```python
condominio.reset_index(drop=True, inplace=True)
condominio.head()
```

```python
condominio.info()
```

```python
import numpy as np
```

```python
condominio['valor_aluguel']=condominio['valor_aluguel'].apply(lambda x: x.replace('$','').replace(',','.').replace('reais','').strip())
condominio['valor_aluguel']=condominio['valor_aluguel'].astype(np.float64)
```

```python
condominio.info()
```

Buscando explicar la organización de la identificación de los apartamentos, durante la creación del conjunto de datos se añadió el texto (blocoAP). Este texto informa que los nombres de los apartamentos están organizados con la letra mayúscula seguida del número del apartamento. Sin embargo, esto no aporta ninguna información a nuestros datos, por lo que resulta interesante eliminar este texto del conjunto de datos.



```python
condominio.head()
```

```python
condominio['apartamento'] = condominio['apartamento'].str.replace(r'\(blocoAP\)', '', regex=True)
condominio.head()
```

Teniendo esto en cuenta, propongo un desafío de procesamiento de datos con el objetivo de analizar el retraso en el pago de la renta en el condominio ficticio de algunos residentes.


```python
condominio.info()
```

```python
datas=[['datas_combinadas_pagamento','datas_de_pagamento']]
```

```python
import pandas as pd
```

```python
for col in datas[0]:
    condominio[col] = pd.to_datetime(condominio[col], format='%d/%m/%Y')
condominio.head()
```

```python
condominio.info()
```

```python
condominio['retraso']=(condominio['datas_de_pagamento']-condominio['datas_combinadas_pagamento']).dt.days
condominio.head()
```

```python
media=condominio.groupby(['apartamento'])['retraso'].mean()
media
```
